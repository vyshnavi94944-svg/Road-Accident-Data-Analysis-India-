# Results & Interpretation

Description:
In this final stage, the key findings from the analysis are summarized and interpreted using the results obtained from the previous stages.

8.1 —Load Final Aggregated Dataset

Description:
Load the aggregated dataset used in the analysis and prepare it for the final interpretation.

```python
import pandas as pd

df = pd.read_csv("/content/aggregated_road_accidents.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

 8.2 — Overall Key Findings

This cell calculates the main totals from the aggregated dataset.

```python
total_accidents = df["accident_count"].sum()
total_fatalities = df["total_fatalities"].sum()
total_severe_injuries = df["total_severe_injuries"].sum()
total_minor_injuries = df["total_minor_injuries"].sum()

print("Overall Key Findings")
print("--------------------")
print("Total Accidents:", total_accidents)
print("Total Fatalities:", total_fatalities)
print("Total Severe Injuries:", total_severe_injuries)
print("Total Minor Injuries:", total_minor_injuries)
```

 8.3 — Description

This cell identifies the State/UT with the highest number of accidents and displays its accident count.

```python
highest_accident_state = df.loc[
    df["accident_count"].idxmax(), "state_ut"
]

highest_accident_count = df["accident_count"].max()

print("State with Highest Accidents")
print("----------------------------")
print("State/UT:", highest_accident_state)
print("Accidents:", highest_accident_count)
```

 8.4 — State with Highest Fatalities

Description: This cell identifies the State/UT with the highest number of fatalities and displays the total fatalities.

```python
highest_fatality_state = df.loc[
    df["total_fatalities"].idxmax(), "state_ut"
]

highest_fatality_count = df["total_fatalities"].max()

print("State with Highest Fatalities")
print("-----------------------------")
print("State/UT:", highest_fatality_state)
print("Fatalities:", highest_fatality_count)
```

8.5 — State with Highest Severe Injuries

Description: This cell identifies the State/UT with the highest number of severe injuries.

```python
highest_severe_state = df.loc[
    df["total_severe_injuries"].idxmax(), "state_ut"
]

highest_severe_count = df["total_severe_injuries"].max()

print("State with Highest Severe Injuries")
print("----------------------------------")
print("State/UT:", highest_severe_state)
print("Severe Injuries:", highest_severe_count)
```

8.6 — State with Highest Minor Injuries

Description: This cell identifies the State/UT with the highest number of minor injuries.

```python
highest_minor_state = df.loc[
    df["total_minor_injuries"].idxmax(), "state_ut"
]

highest_minor_count = df["total_minor_injuries"].max()

print("State with Highest Minor Injuries")
print("---------------------------------")
print("State/UT:", highest_minor_state)
print("Minor Injuries:", highest_minor_count)
```

8.7 — Total Accident Outcomes

Description: This cell calculates the total number of injuries and fatalities together to understand the overall accident impact.

```python
total_injuries = (
    df["total_severe_injuries"].sum() +
    df["total_minor_injuries"].sum()
)

total_outcomes = (
    total_injuries +
    df["total_fatalities"].sum()
)

print("Total Accident Outcomes")
print("-----------------------")
print("Total Injuries:", total_injuries)
print("Total Fatalities:", df["total_fatalities"].sum())
print("Total Outcomes:", total_outcomes)
```

8.8 — Key Observations

Description: This cell summarizes the main findings identified from the accident data.

```python
print("KEY OBSERVATIONS")
print("----------------")

print("• Maharashtra has the highest number of accidents.")
print("• Maharashtra also has the highest number of fatalities.")
print("• Severe and minor injuries form a major part of accident outcomes.")
print("• Accident patterns vary across different States/UTs.")
```

8.9 — Accident Impact Summary

Description: This cell combines the major accident outcomes into one simple summary.

```python
print("ACCIDENT IMPACT SUMMARY")
print("-----------------------")

print("Total Accidents:", df["accident_count"].sum())
print("Total Fatalities:", df["total_fatalities"].sum())
print("Total Injuries:",
      df["total_severe_injuries"].sum() +
      df["total_minor_injuries"].sum())

print("\nHighest Accident State:", highest_accident_state)
print("Highest Fatality State:", highest_fatality_state)
```

Description: This final cell presents the overall conclusion of the road accident analysis project.

```python
print("FINAL PROJECT CONCLUSION")
print("------------------------")

print("The analysis identified differences in road accident")
print("patterns and outcomes across States/UTs.")

print("Maharashtra recorded the highest number of accidents")
print("and fatalities in the dataset.")

print("The analysis also showed significant numbers of")
print("severe and minor injuries.")

print("Overall, the project helps summarize accident patterns")
print("and their major outcomes using data analysis.")
```
