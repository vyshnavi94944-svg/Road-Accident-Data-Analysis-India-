# DATA ACQUISITION & FILTERING

```python
# STAGE 2: DATA ACQUISITION & FILTERING
# Acquire the required data from the dataset

import pandas as pd

file_path = "/content/india_road_accidents.csv"
df = pd.read_csv(file_path)

print("Dataset acquired successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nAvailable Columns:")
print(df.columns.tolist())
```

```python

# Select the important columns needed for analysis

required_columns = [
    "state_ut",
    "district",
    "road_type",
    "weather",
    "severity",
    "primary_cause",
    "fatalities",
    "severe_injuries",
    "minor_injuries"
]

acquired_data = df[required_columns]

print("Required data acquired successfully!")
print("Rows:", acquired_data.shape[0])
print("Columns:", acquired_data.shape[1])

acquired_data.head()
```

```python
# Create a dataset containing only accident records needed for further analysis

analysis_data = acquired_data[
    [
        "state_ut",
        "district",
        "road_type",
        "weather",
        "severity",
        "primary_cause",
        "fatalities",
        "severe_injuries",
        "minor_injuries"
    ]
].copy()

print("Analysis data acquired successfully!")
print("Rows:", len(analysis_data))
print("Columns:", len(analysis_data.columns))

analysis_data.head()
```

```python
# Create a smaller dataset for accident outcome analysis

outcome_data = analysis_data[
    [
        "severity",
        "fatalities",
        "severe_injuries",
        "minor_injuries"
    ]
].copy()

print("Outcome data extracted successfully!")
print("Rows:", len(outcome_data))
print("Columns:", len(outcome_data.columns))

outcome_data.head()
```

```python

# Filter the data for fatal accidents

fatal_accidents = acquired_data[acquired_data["severity"] == "Fatal"]

print("Fatal accidents:", len(fatal_accidents))
fatal_accidents.head()
```

```python
# Filter by road type

national_highway = acquired_data[
    acquired_data["road_type"] == "National Highway"
]

print("National Highway accidents:", len(national_highway))
national_highway.head()
```

```python
# Filter by weather condition

heavy_rain = acquired_data[
    acquired_data["weather"] == "Heavy Rain"
]

print("Heavy Rain accidents:", len(heavy_rain))
heavy_rain.head()
```

```python
# Filter by a specific state

telangana_accidents = acquired_data[
    acquired_data["state_ut"] == "Telangana"
]

print("Telangana accidents:", len(telangana_accidents))
telangana_accidents.head()
```

```python
# Filter accidents by both state and severity

telangana_fatal = acquired_data[
    (acquired_data["state_ut"] == "Telangana") &
    (acquired_data["severity"] == "Fatal")
]

print("Fatal accidents in Telangana:", len(telangana_fatal))
telangana_fatal.head()
```

```python
# Filter accidents with high-severity outcomes

serious_accidents = acquired_data[
    acquired_data["severity"].isin(["Fatal", "Grievous Injury"])
]

print("Serious accidents:", len(serious_accidents))
serious_accidents.head()
```

```python
# Filter accidents caused by overspeeding

overspeeding_accidents = acquired_data[
    acquired_data["primary_cause"] == "Overspeeding"
]

print("Overspeeding accidents:", len(overspeeding_accidents))
overspeeding_accidents.head()
```

```python
# Filter only fatal accidents

fatal_data = acquired_data[
    acquired_data["severity"] == "Fatal"
]

print("Fatal accidents:", len(fatal_data))
fatal_data.head()
```

```python
# Filter accidents with at least one fatality

fatality_records = acquired_data[
    acquired_data["fatalities"] > 0
]

print("Accidents with fatalities:", len(fatality_records))
fatality_records.head()
```

```python
# Filter accidents with severe injuries

severe_injury_records = acquired_data[
    acquired_data["severe_injuries"] > 0
]

print("Accidents with severe injuries:", len(severe_injury_records))
severe_injury_records.head()
```
