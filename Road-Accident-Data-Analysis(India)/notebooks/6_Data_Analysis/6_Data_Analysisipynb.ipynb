# Data Analysis

Description:
In this stage, the aggregated road accident data is analyzed using statistical calculations and comparisons to identify important patterns in accidents, fatalities, injuries, severity, road types, weather conditions, and accident causes.

6.1 — Load the Aggregated Dataset

Description:
Load the aggregated dataset generated in Stage 5 and verify its structure before performing the analysis.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/aggregated_road_accidents.csv")

print("Aggregated dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nDataset preview:")
print(df.head())
```

6.2 — Calculate Overall Accident and Casualty Totals

Description:
Calculate the total number of accidents, fatalities, severe injuries, and minor injuries represented in the aggregated dataset.

```python
total_accidents = df["accident_count"].sum()
total_fatalities = df["total_fatalities"].sum()
total_severe_injuries = df["total_severe_injuries"].sum()
total_minor_injuries = df["total_minor_injuries"].sum()

print("Overall Analysis")
print("----------------")
print("Total Accidents:", total_accidents)
print("Total Fatalities:", total_fatalities)
print("Total Severe Injuries:", total_severe_injuries)
print("Total Minor Injuries:", total_minor_injuries)
```

6.3 — Calculate Average Accidents per State/UT

Description:
Calculate the average number of recorded accidents per State/UT to understand the overall distribution of accidents across the states in the dataset.

```python
average_accidents = df["accident_count"].mean()

print("Average accidents per State/UT:", round(average_accidents, 2))
```

6.4 — Identify the State/UT with the Highest Number of Accidents

Description:
Identify the State/UT with the highest recorded number of accidents in the dataset.

```python
highest_accident_state = df.loc[
    df["accident_count"].idxmax()
]

print("State/UT with the highest number of accidents:")
print("State/UT:", highest_accident_state["state_ut"])
print("Accidents:", highest_accident_state["accident_count"])
```

6.5 — Identify the State/UT with the Highest Fatalities

Description:
Identify the State/UT with the highest total number of fatalities recorded in the aggregated dataset.

```python
highest_fatality_state = df.loc[
    df["total_fatalities"].idxmax()
]

print("State/UT with the highest fatalities:")
print("State/UT:", highest_fatality_state["state_ut"])
print("Fatalities:", highest_fatality_state["total_fatalities"])
```

6.6 — Analyze Injury Distribution

Description:
Compare the total severe and minor injuries to determine how injuries are distributed across the accident records.

```python
total_injuries = (
    df["total_severe_injuries"].sum()
    + df["total_minor_injuries"].sum()
)

severe_percentage = (
    df["total_severe_injuries"].sum() / total_injuries
) * 100

minor_percentage = (
    df["total_minor_injuries"].sum() / total_injuries
) * 100

print("Injury Distribution")
print("--------------------")
print("Total Injuries:", total_injuries)
print("Severe Injuries:", df["total_severe_injuries"].sum())
print("Minor Injuries:", df["total_minor_injuries"].sum())
print("Severe Injury Percentage:", round(severe_percentage, 2), "%")
print("Minor Injury Percentage:", round(minor_percentage, 2), "%")
```

6.7 — Calculate Accident Share by State/UT

Description:
Calculate the percentage contribution of each State/UT to the total number of recorded accidents.

```python
df["accident_percentage"] = (
    df["accident_count"] / df["accident_count"].sum()
) * 100

print("Accident Share by State/UT:")
print(
    df[["state_ut", "accident_count", "accident_percentage"]]
    .sort_values("accident_percentage", ascending=False)
)
```

6.8 — Calculate Fatality Rate per Accident

Description:
Calculate the average number of fatalities per accident for each State/UT. This helps compare fatality levels relative to the number of accidents.

```python
df["fatalities_per_accident"] = (
    df["total_fatalities"] / df["accident_count"]
)

print("Fatalities per Accident by State/UT:")
print(
    df[["state_ut", "accident_count", "total_fatalities",
        "fatalities_per_accident"]]
    .sort_values("fatalities_per_accident", ascending=False)
)
```

6.9 — Analyze Injury Burden by State/UT

Description:
Calculate the total injuries for each State/UT by combining severe and minor injuries, allowing comparison of the overall injury burden.

```python
df["total_injuries"] = (
    df["total_severe_injuries"]
    + df["total_minor_injuries"]
)

injury_analysis = df[
    ["state_ut", "total_severe_injuries",
     "total_minor_injuries", "total_injuries"]
].sort_values("total_injuries", ascending=False)

print("Injury Burden by State/UT:")
print(injury_analysis)
```

6.10 — Compare Fatalities and Injuries

Description:
Compare fatalities, severe injuries, and minor injuries across State/UTs to identify how accident outcomes differ between locations.

```python
outcome_comparison = df[
    ["state_ut", "total_fatalities",
     "total_severe_injuries", "total_minor_injuries"]
].copy()

outcome_comparison["total_casualties"] = (
    outcome_comparison["total_fatalities"]
    + outcome_comparison["total_severe_injuries"]
    + outcome_comparison["total_minor_injuries"]
)

print("Accident Outcome Comparison:")
print(
    outcome_comparison.sort_values(
        "total_casualties",
        ascending=False
    )
)
```

6.11 — Identify the State/UT with the Highest Total Casualties

Description:
Identify the State/UT with the highest combined number of fatalities and injuries

```python
df["total_casualties"] = (
    df["total_fatalities"]
    + df["total_severe_injuries"]
    + df["total_minor_injuries"]
)

highest_casualty_state = df.loc[
    df["total_casualties"].idxmax()
]

print("State/UT with the highest total casualties:")
print("State/UT:", highest_casualty_state["state_ut"])
print("Total Casualties:", highest_casualty_state["total_casualties"])
```

6.12 — Final Analysis Summary

Description:
Summarize the main accident, fatality, and injury findings from the dataset.

```python
print("FINAL ANALYSIS SUMMARY")
print("----------------------")

print("Total Accidents:", df["accident_count"].sum())
print("Total Fatalities:", df["total_fatalities"].sum())
print("Total Severe Injuries:", df["total_severe_injuries"].sum())
print("Total Minor Injuries:", df["total_minor_injuries"].sum())

print("Highest Accident State:",
      df.loc[df["accident_count"].idxmax(), "state_ut"])

print("Highest Fatality State:",
      df.loc[df["total_fatalities"].idxmax(), "state_ut"])
```
