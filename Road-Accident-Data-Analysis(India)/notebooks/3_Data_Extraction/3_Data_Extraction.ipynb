# Data Extraction
In this stage, relevant information is extracted from the road accident dataset and organized into smaller datasets based on accident causes, locations, road conditions, and accident outcomes. This makes the data easier to use in the later stages of analysis.

3.1 Import Libraries and Load Dataset

The required Python libraries are imported and the road accident dataset is loaded into a DataFrame for further extraction.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/india_road_accidents.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

3.2 Extract Accident Cause Information

Description: This cell extracts the main accident-related information, including the primary cause, weather condition, and road type.

```python
cause_data = df[["primary_cause", "weather", "road_type"]].copy()

print("Accident cause information extracted successfully!")
print(cause_data.head())
```

3.3 Extract Location Information

Description: This cell extracts the location-related information, including the state/UT, district, and road type of each accident.

```python
location_data = df[["state_ut", "district", "road_type"]].copy()

print("Location information extracted successfully!")
print(location_data.head())
```

3.4 Extract Accident Outcome Information

Description: This cell extracts the accident outcome information, including severity, fatalities, severe injuries, and minor injuries.

```python
outcome_data = df[
    ["severity", "fatalities", "severe_injuries", "minor_injuries"]
].copy()

print("Accident outcome information extracted successfully!")
print(outcome_data.head())
```

3.5 Create a Combined Extracted Dataset

Description: This cell combines the extracted location, cause, and outcome information into one dataset containing the key accident details for further analysis.

```python
extracted_data = df[
    [
        "state_ut",
        "district",
        "road_type",
        "weather",
        "severity",
        "primary_cause",
        "fatalities",
        "severe_injuries",
        "minor_injuries"
    ]
].copy()

print("Combined extracted dataset created successfully!")
print("Rows:", extracted_data.shape[0])
print("Columns:", extracted_data.shape[1])
print(extracted_data.head())
```

3.6 Extract Accident Records by Severity

Description: This cell separates the accident records into three simple groups based on severity: Fatal, Grievous Injury, and Minor Injury.

```python
fatal_data = df[df["severity"] == "Fatal"]

grievous_data = df[df["severity"] == "Grievous Injury"]

minor_data = df[df["severity"] == "Minor Injury"]

print("Fatal accidents:", len(fatal_data))
print("Grievous Injury accidents:", len(grievous_data))
print("Minor Injury accidents:", len(minor_data))
```

3.7 Extract Accidents by Road Type

Description: This cell separates the accident records into groups based on the type of road where the accidents occurred.

```python
national_highway = df[df["road_type"] == "National Highway"]

state_highway = df[df["road_type"] == "State Highway"]

city_road = df[df["road_type"] == "City Road"]

print("National Highway accidents:", len(national_highway))
print("State Highway accidents:", len(state_highway))
print("City Road accidents:", len(city_road))
```

3.8 Extract Accidents by Weather Condition

Description: This cell extracts accident records based on different weather conditions to understand how accidents are distributed across weather situations.

```python
weather_clear = df[df["weather"] == "Clear"]

weather_rain = df[df["weather"] == "Heavy Rain"]

weather_fog = df[df["weather"] == "Dense Fog"]

weather_heat = df[df["weather"] == "Extreme Heat"]

print("Clear weather accidents:", len(weather_clear))
print("Heavy Rain accidents:", len(weather_rain))
print("Dense Fog accidents:", len(weather_fog))
print("Extreme Heat accidents:", len(weather_heat))
```

3.9 Extract Accidents by State/UT

Description: This cell extracts accident records for each state or union territory, helping us organize the dataset based on accident location.

```python
state_counts = df["state_ut"].value_counts()

print("Accidents by State/UT:")
print(state_counts)
```

3.10 Extract Accidents by Primary Cause

Description: This cell extracts the number of accident records associated with each primary cause. This helps organize the data according to the reported causes of accidents.

```python
cause_counts = df["primary_cause"].value_counts()

print("Accidents by Primary Cause:")
print(cause_counts)
```

3.11 Extract Accidents with Fatalities

Description: This cell extracts only those accident records where at least one fatality was reported.

```python
fatality_data = df[df["fatalities"] > 0]

print("Accidents with fatalities:", len(fatality_data))
print(fatality_data.head())
```

3.12 Extract Accidents with Severe Injuries

Description: This cell extracts accident records where at least one severe injury was reported.

```python
severe_injury_data = df[df["severe_injuries"] > 0]

print("Accidents with severe injuries:", len(severe_injury_data))
print(severe_injury_data.head())
```

3.13 Extract Accidents with Minor Injuries

Description: This cell extracts accident records where at least one minor injury was reported.

```python
minor_injury_data = df[df["minor_injuries"] > 0]

print("Accidents with minor injuries:", len(minor_injury_data))
print(minor_injury_data.head())
```

3.14 Extract Accidents by District

Description: This cell extracts the number of accident records for each district, helping organize the accident data according to specific locations.

```python
district_counts = df["district"].value_counts()

print("Accidents by District:")
print(district_counts)
```

3.15 Extract Accidents Caused by Overspeeding

Description: This cell extracts accident records where overspeeding was reported as the primary cause.

```python
overspeeding_data = df[df["primary_cause"] == "Overspeeding"]

print("Overspeeding-related accidents:", len(overspeeding_data))
print(overspeeding_data.head())
```

3.16 Create Final Extracted Dataset

Description: This cell creates the final extracted dataset containing all important accident details. It also verifies the number of records and columns extracted before moving to the data validation and cleaning stage.

```python
extracted_data = df[
    [
        "state_ut",
        "district",
        "road_type",
        "weather",
        "severity",
        "primary_cause",
        "fatalities",
        "severe_injuries",
        "minor_injuries"
    ]
].copy()

print("Final extracted dataset created successfully!")
print("Rows:", extracted_data.shape[0])
print("Columns:", extracted_data.shape[1])

print("\nColumn names:")
print(extracted_data.columns.tolist())

print("\nFirst 5 records:")
print(extracted_data.head())
```

Description: This cell saves the final extracted dataset as a CSV file so it can be used as the input dataset for Stage 4.

```python
extracted_data.to_csv(
    "/content/extracted_road_accidents.csv",
    index=False
)

print("Extracted dataset saved successfully!")
```
