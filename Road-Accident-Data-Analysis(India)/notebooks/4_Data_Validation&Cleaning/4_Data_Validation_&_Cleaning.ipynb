# Data Validation & Cleaning

Description: In this stage, the extracted road accident dataset is checked for missing values, duplicate records, inconsistent text values, incorrect data types, and invalid numerical values. Necessary cleaning steps are applied to improve the quality and reliability of the data.

4.1 Import Libraries and Load Dataset

Description: The required libraries are imported and the Stage 3 extracted dataset is loaded into a new DataFrame for validation and cleaning.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/extracted_road_accidents.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

4.2 Check Missing Values

Description: This cell checks every column for missing values. Missing values can affect the accuracy of later analysis, so they need to be identified before cleaning the dataset.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

4.3 Check Duplicate Records

Description: This cell checks whether the dataset contains any completely duplicated rows. Duplicate records can lead to incorrect counts and affect the accuracy of the analysis

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

4.4 Check Data Types

Description: This cell checks the data type of each column to make sure text columns are stored as text and numerical columns are stored as numbers.

```python
print("Data types of each column:")
print(df.dtypes)
```

4.5 Check Unique Values in Categorical Columns

Description: This cell checks the different values present in important categorical columns. This helps identify inconsistent or unexpected entries before cleaning.

```python
print("Unique values in Severity:")
print(df["severity"].unique())

print("\nUnique values in Road Type:")
print(df["road_type"].unique())

print("\nUnique values in Weather:")
print(df["weather"].unique())
```

4.6 Check Unique Values in Location and Cause Columns

Description: This cell checks the unique values in the state/UT and primary cause columns. This helps identify differences in capitalization, spelling, or formatting that may need to be standardized during cleaning.

```python
print("Unique State/UT values:")
print(df["state_ut"].unique())

print("\nUnique Primary Cause values:")
print(df["primary_cause"].unique())
```

4.7 Check Numerical Values

Description: This cell checks the minimum and maximum values of fatalities and injury columns. This helps identify any negative or unusually invalid values before cleaning.

```python
print("Fatalities - Minimum:", df["fatalities"].min())
print("Fatalities - Maximum:", df["fatalities"].max())

print("Severe Injuries - Minimum:", df["severe_injuries"].min())
print("Severe Injuries - Maximum:", df["severe_injuries"].max())

print("Minor Injuries - Minimum:", df["minor_injuries"].min())
print("Minor Injuries - Maximum:", df["minor_injuries"].max())
```

4.8 Check for Negative Values

Description: This cell checks whether any accident record contains negative values for fatalities or injuries. Negative values are not valid for these fields and would need to be corrected during cleaning.

```python
print("Negative fatalities:", (df["fatalities"] < 0).sum())
print("Negative severe injuries:", (df["severe_injuries"] < 0).sum())
print("Negative minor injuries:", (df["minor_injuries"] < 0).sum())
```

4.9 Check for Invalid Severity Values

Description: This cell checks whether the severity column contains only the expected accident severity categories.

```python
valid_severity = ["Fatal", "Grievous Injury", "Minor Injury"]

invalid_severity = df[~df["severity"].isin(valid_severity)]

print("Number of invalid severity records:", len(invalid_severity))
print(invalid_severity[["severity"]].drop_duplicates())
```

4.10 Check for Invalid Road Type Values

Description: This cell checks whether the road type column contains only the expected road categories. This helps identify any unexpected values that may need cleaning.

```python
valid_road_types = [
    "National Highway",
    "State Highway",
    "City Road"
]

invalid_road_types = df[~df["road_type"].isin(valid_road_types)]

print("Number of invalid road type records:", len(invalid_road_types))
print(invalid_road_types[["road_type"]].drop_duplicates())
```

4.11 Check for Invalid Weather Values

Description: This cell checks whether the weather column contains only the expected weather categories.

```python
valid_weather = ["Clear", "Heavy Rain", "Dense Fog", "Extreme Heat"]

invalid_weather = df[~df["weather"].isin(valid_weather)]

print("Invalid weather records:", len(invalid_weather))
print(invalid_weather[["weather"]].drop_duplicates())
```

4.12 Final Validation Summary

Description: This cell gives a final summary of the main validation checks before the cleaning process begins.

```python
print("FINAL VALIDATION SUMMARY")
print("------------------------")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
print("Negative numerical values:",
      ((df["fatalities"] < 0) |
       (df["severe_injuries"] < 0) |
       (df["minor_injuries"] < 0)).sum())
```

4.13 Standardize Text Values

Description: This cell standardizes text values by removing extra spaces and making the formatting consistent. This is useful for fixing entries such as DENSE FOG and Dense Fog.

```python
text_columns = [
    "state_ut",
    "district",
    "road_type",
    "weather",
    "severity",
    "primary_cause"
]

for column in text_columns:
    df[column] = df[column].str.strip()

df["weather"] = df["weather"].str.title()
df["severity"] = df["severity"].str.title()
df["road_type"] = df["road_type"].str.title()

print("Text values standardized successfully!")
```

4.14 Standardize State Names

Description: This cell standardizes state and UT names by removing extra spaces and using consistent capitalization.

```python
df["state_ut"] = df["state_ut"].str.strip().str.title()

print("State/UT names standardized successfully!")
print(df["state_ut"].unique())
```

4.15 Remove Duplicate Records

Description: This cell removes any completely duplicated accident records to prevent the same accident from being counted more than once.

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Duplicate records removed:", before - after)
print("Rows after cleaning:", after)
```

4.16 Handle Missing Values

Description: This cell checks for missing values and fills missing text values with "Unknown" so that important accident records are not unnecessarily removed.

```python
text_columns = [
    "state_ut",
    "district",
    "road_type",
    "weather",
    "severity",
    "primary_cause"
]

for column in text_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing text values handled successfully!")
print("Remaining missing values:", df.isnull().sum().sum())
```

4.17 Convert Numerical Columns to Correct Data Type

Description: This cell converts the accident and injury columns into numeric format so they can be safely used for calculations and analysis.

```python
numeric_columns = [
    "fatalities",
    "severe_injuries",
    "minor_injuries"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("Numerical columns converted successfully!")
print(df[numeric_columns].dtypes)
```

4.18 Handle Invalid Numerical Values

Description: This cell checks for negative values in fatalities and injury columns and replaces any invalid negative values with 0.

```python
numeric_columns = [
    "fatalities",
    "severe_injuries",
    "minor_injuries"
]

for column in numeric_columns:
    df.loc[df[column] < 0, column] = 0

print("Invalid negative values handled successfully!")
print("Remaining negative values:",
      (df[numeric_columns] < 0).sum().sum())
```

4.19 Finalize Cleaned Dataset

Description: This cell performs a final check of the cleaned dataset and saves it as a new CSV file for use in Stage 5.

```python
print("Final cleaned dataset:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Remaining missing values:", df.isnull().sum().sum())
print("Remaining duplicates:", df.duplicated().sum())

df.to_csv("/content/cleaned_road_accidents.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
