# Data Visualization

Description:
In this stage, the analyzed road accident data is presented using charts and graphs to visually identify patterns in accidents, fatalities, injuries, and accident distribution.

7.1 — Load the Aggregated Dataset

Description:
Load the aggregated dataset from Stage 5 and prepare it for visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/aggregated_road_accidents.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

7.2 — Bar Chart: Accidents by State/UT

Description:
Create a bar chart to compare the number of recorded accidents across different State/UT categories.

```python
plt.figure(figsize=(12, 6))

plt.bar(df["state_ut"], df["accident_count"])

plt.xlabel("State/UT")
plt.ylabel("Number of Accidents")
plt.title("Accidents by State/UT")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig("accidents_by_state.png", bbox_inches="tight")
plt.show()
```

7.3 — Pie Chart: Accident Distribution by State/UT

Description:
Create a pie chart to show how the total recorded accidents are distributed among the State/UT categories.

```python
plt.figure(figsize=(8, 8))

plt.pie(
    df["accident_count"],
    labels=df["state_ut"],
    autopct="%1.1f%%"
)

plt.title("Accident Distribution by State/UT")
plt.savefig("accident_distribution_by_state.png", bbox_inches="tight")
plt.show()
```

7.4 — Bar Chart: Fatalities by State/UT

Description:
Create a bar chart to compare the total number of fatalities recorded in each State/UT.

```python
plt.figure(figsize=(12, 6))

plt.bar(df["state_ut"], df["total_fatalities"])

plt.xlabel("State/UT")
plt.ylabel("Total Fatalities")
plt.title("Fatalities by State/UT")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig("fatalities_by_state.png", bbox_inches="tight")
plt.show()
```

7.5 — Scatter Plot: Accidents vs Fatalities

Description:
Create a scatter plot to examine the relationship between the number of accidents and fatalities across State/UTs.

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["accident_count"],
    df["total_fatalities"]
)

plt.xlabel("Number of Accidents")
plt.ylabel("Total Fatalities")
plt.title("Accidents vs Fatalities")
plt.savefig("accidents_vs_fatalities.png", bbox_inches="tight")
plt.show()
```

7.6 — Histogram: Distribution of Accidents

Description:
Create a histogram to show how the number of accidents is distributed across the State/UTs.

```python
plt.figure(figsize=(8, 6))

plt.hist(df["accident_count"], bins=8)

plt.xlabel("Number of Accidents")
plt.ylabel("Number of States/UTs")
plt.title("Distribution of Accidents")
plt.savefig("accident_distribution.png", bbox_inches="tight")
plt.show()
```

7.7 — Box Plot: Fatalities Distribution

Description:
Create a box plot to show the distribution of fatalities across the State/UTs and identify the spread of the data.

```python
plt.figure(figsize=(6, 5))

plt.boxplot(df["total_fatalities"])

plt.ylabel("Total Fatalities")
plt.title("Distribution of Fatalities")
plt.savefig("distribution_of_fatalities.png", bbox_inches="tight")
plt.show()
```

7.8 — Stacked Bar Chart: Injury Comparison

Description:
Compare severe and minor injuries across State/UTs using a simple stacked bar chart.

```python
df.plot(
    x="state_ut",
    y=["total_severe_injuries", "total_minor_injuries"],
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title("Severe and Minor Injuries by State/UT")
plt.xlabel("State/UT")
plt.ylabel("Injuries")
plt.xticks(rotation=90)
plt.savefig("severe_minor_injuries_by_state.png", bbox_inches="tight")
plt.show()
```

7.9 — Scatter Plot: Accidents vs Total Injuries

Description:
Show the relationship between the number of accidents and total injuries across State/UTs.

```python
df["total_injuries"] = (
    df["total_severe_injuries"] +
    df["total_minor_injuries"]
)

plt.scatter(df["accident_count"], df["total_injuries"])

plt.xlabel("Accidents")
plt.ylabel("Total Injuries")

plt.title("Accidents vs Total Injuries")

plt.savefig("accidents_vs_total_injuries.png", bbox_inches="tight")

plt.show()
```

7.10 — Final Visualization Summary

Description:
Display a simple summary of the main visualized accident outcomes.

```python
print("FINAL VISUALIZATION SUMMARY")
print("---------------------------")

print("Total Accidents:", df["accident_count"].sum())
print("Total Fatalities:", df["total_fatalities"].sum())
print("Total Severe Injuries:", df["total_severe_injuries"].sum())
print("Total Minor Injuries:", df["total_minor_injuries"].sum())

highest_accident_state = df.loc[
    df["accident_count"].idxmax(), "state_ut"
]

highest_fatality_state = df.loc[
    df["total_fatalities"].idxmax(), "state_ut"
]

print("\nCONCLUSION FROM VISUALIZATIONS")
print("-------------------------------")
print("• Accident counts vary across different States/UTs.")
print("• Maharashtra has the highest number of accidents in the dataset.")
print("• Maharashtra also has the highest number of fatalities.")
print("• Severe and minor injuries contribute significantly to the overall accident outcomes.")
print("• The scatter plots show the relationship between accident counts and accident outcomes.")
print("• Overall, the visualizations help identify differences in accident patterns and their outcomes across States/UTs.")
```
