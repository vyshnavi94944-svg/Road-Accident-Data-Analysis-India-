# Data Aggregation & Representation
Description:
In this stage, the cleaned road accident data is grouped and summarized to identify accident patterns across states, districts, severity levels, road types, weather conditions, and primary causes. Cross-tabulations are also created to represent relationships between important accident factors.

5.1 — Load the Cleaned Dataset

Description:
Load the cleaned dataset generated in Stage 4 and verify its size before performing aggregation.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/cleaned_road_accidents.csv")

print("Cleaned dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

5.2 — Aggregate Accidents by State/UT

Description:
Group the cleaned data by State/UT and calculate the total number of accidents recorded in each state.

```python
state_accidents = df.groupby("state_ut").size().reset_index(name="accident_count")

print("Accidents by State/UT:")
print(state_accidents)
```

5.3 — Aggregate Accidents by Severity

Description:
Group the accident records according to their severity level and calculate the total number of accidents in each category.

```python
severity_summary = df.groupby("severity").size().reset_index(name="accident_count")

print("Accidents by Severity:")
print(severity_summary)
```

5.4 — Aggregate Accidents by Road Type

Description:
Group the accident records by road type and calculate the total number of accidents for each road category.

```python
road_summary = df.groupby("road_type").size().reset_index(name="accident_count")

print("Accidents by Road Type:")
print(road_summary)
```

5.5 — Aggregate Accidents by Weather

Description:
Group the accident records by weather condition and calculate the total number of accidents recorded under each condition.

```python
weather_summary = df.groupby("weather").size().reset_index(name="accident_count")

print("Accidents by Weather Condition:")
print(weather_summary)
```

5.6 — Aggregate Accidents by Primary Cause

Description:
Group the accident records by their primary cause and calculate the total number of accidents associated with each cause.

```python
cause_summary = df.groupby("primary_cause").size().reset_index(name="accident_count")

print("Accidents by Primary Cause:")
print(cause_summary)
```

5.7 — Aggregate Fatalities and Injuries by State/UT

Description:
Group the data by State/UT and calculate the total fatalities, severe injuries, and minor injuries recorded in each state.

```python
state_outcomes = df.groupby("state_ut")[
    ["fatalities", "severe_injuries", "minor_injuries"]
].sum().reset_index()

print("Fatalities and Injuries by State/UT:")
print(state_outcomes)
```

5.8 — Severity × Road Type Representation

Description:
Create a cross-tabulation showing how accident severity is distributed across different road types.

```python
severity_road_table = pd.crosstab(
    df["road_type"],
    df["severity"]
)

print("Severity by Road Type:")
print(severity_road_table)
```

5.9 — Weather × Severity Representation

Description:
Create a cross-tabulation to represent the distribution of accident severity under different weather conditions.

```python
weather_severity_table = pd.crosstab(
    df["weather"],
    df["severity"]
)

print("Severity by Weather Condition:")
print(weather_severity_table)
```

5.10 — Final Aggregated Summary

Description:
Create a final state-wise summary combining accident counts, fatalities, and injuries to provide a compact representation of the aggregated dataset.

```python
final_summary = df.groupby("state_ut").agg(
    accident_count=("state_ut", "size"),
    total_fatalities=("fatalities", "sum"),
    total_severe_injuries=("severe_injuries", "sum"),
    total_minor_injuries=("minor_injuries", "sum")
).reset_index()

print("Final Aggregated Summary:")
print(final_summary)

final_summary.to_csv(
    "/content/aggregated_road_accidents.csv",
    index=False
)

print("\nAggregated summary saved successfully!")
```
